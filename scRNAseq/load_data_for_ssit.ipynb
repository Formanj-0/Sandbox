```python
import pandas as pd
```

```python
rawCount_path = r'GSE141834_scRNAseq_rawCounts.txt'
normalizedCount_path = r'GSE141834_scRNAseq_seuratV3_normalized.txt'

raw_counts_df = pd.read_csv(rawCount_path, sep='\t', index_col=0)
normalized_counts_df = pd.read_csv(normalizedCount_path, sep='\t', index_col=0)

raw_counts_df = raw_counts_df.T
normalized_counts_df = normalized_counts_df.T
```

```python
raw_counts_df['time'] = raw_counts_df.index.str.split('.').str[1]
normalized_counts_df['time'] = normalized_counts_df.index.str.split('.').str[1]

raw_counts_df['time'] = raw_counts_df['time'].astype(int)
normalized_counts_df['time'] = normalized_counts_df['time'].astype(int)
```

```python
raw_counts_df.to_csv('Dex_scRNAseq_raw_counts.csv')
normalized_counts_df.to_csv('Dex_scRNAseq_normalized_counts.csv')
```
